```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../bluestock_mf.db")
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT scheme_name,
       aum_crore
FROM fact_performance
ORDER BY aum_crore DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT amfi_code,
       ROUND(AVG(nav),2) AS avg_nav
FROM fact_nav
GROUP BY amfi_code
ORDER BY avg_nav DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT transaction_type,
       COUNT(*) AS total_transactions
FROM fact_transaction
GROUP BY transaction_type;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT state,
       ROUND(SUM(amount_inr),2) AS total_amount
FROM fact_transaction
GROUP BY state
ORDER BY total_amount DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT investor_id,
       ROUND(SUM(amount_inr),2) AS invested_amount
FROM fact_transaction
GROUP BY investor_id
ORDER BY invested_amount DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT kyc_status,
       COUNT(*) AS total
FROM fact_transaction
GROUP BY kyc_status;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT scheme_name,
       sharpe_ratio
FROM fact_performance
ORDER BY sharpe_ratio DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT scheme_name,
       return_3yr_pct
FROM fact_performance
ORDER BY return_3yr_pct DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
ROUND(AVG(alpha),4) AS avg_alpha,
ROUND(AVG(beta),4) AS avg_beta
FROM fact_performance;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT category,
       COUNT(*) AS fund_count
FROM dim_fund
GROUP BY category
ORDER BY fund_count DESC;
"""

pd.read_sql(query, conn)
```
